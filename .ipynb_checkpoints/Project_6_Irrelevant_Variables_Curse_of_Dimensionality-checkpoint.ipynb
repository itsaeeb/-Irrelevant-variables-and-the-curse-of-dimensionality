## 1. Theoretical Analysis

### 1.1 Data-Generating Model and Comparability Across $r$

Each observation is $x = (z, u) \in \mathbb{R}^d$, with binary label $y \in \{0, 1\}$ and equal class priors $P(y=0) = P(y=1) = 0.5$:
- **Informative coordinates** $z \in \mathbb{R}^r$:
  $$z \mid (y = 0) \sim \mathcal{N}(-\delta \mathbf{1}_r, \sigma_z^2 I_r), \qquad z \mid (y = 1) \sim \mathcal{N}(+\delta \mathbf{1}_r, \sigma_z^2 I_r)$$
- **Irrelevant coordinates** $u \in \mathbb{R}^{d-r}$:
  $$u \sim \mathcal{N}(0, \sigma_u^2 I_{d-r}), \qquad u \perp\!\!\!\perp (z, y)$$

Because $u$ is independent of $(z, y)$, the true log-posterior odds depend only on $z$:
$$\log \frac{P(y = 1 \mid x)}{P(y = 0 \mid x)} = \log \frac{P(y = 1 \mid z)}{P(y = 0 \mid z)} = \frac{2\delta}{\sigma_z^2} \sum_{j=1}^r z_j$$

To prevent higher-$r$ problems from becoming artificially easier when $r$ increases, we scale the per-coordinate shift as:
$$\delta = \frac{\mu_0}{\sqrt{r}} \sigma_z \qquad (\mu_0 = 1.2)$$

Defining the standardized projection $s(z) = \frac{1}{\sigma_z \sqrt{r}} \sum_{j=1}^r z_j$, we have $s(z) \mid (y=0) \sim \mathcal{N}(-\mu_0, 1)$ and $s(z) \mid (y=1) \sim \mathcal{N}(+\mu_0, 1)$, so $P(y=1 \mid z) = \sigma(2\mu_0 s(z))$. Consequently, both the **Bayes optimal error rate** $\Phi(-\mu_0) = \Phi(-1.2) \approx 11.51\%$ and the **Bayes optimal log-loss** remain identical across every informative dimension $r$ and every feature scale $\sigma_z$.

---
### 1.2 Moments of Squared Distances and Distance Concentration

For two independent observations $x = (z, u)$ and $x' = (z', u')$, the squared Euclidean distance decomposes additively:
$$\|x - x'\|^2 = \underbrace{\|z - z'\|^2}_{\text{signal}} + \underbrace{\|u - u'\|^2}_{\text{noise}}$$

Since $u_j - u'_j \sim \mathcal{N}(0, 2\sigma_u^2)$ independently for $j = 1, \dots, d - r$:
$$\frac{\|u - u'\|^2}{2\sigma_u^2} \sim \chi^2(d - r) \implies \mathbb{E}\!\left[\|u - u'\|^2\right] = 2(d - r)\sigma_u^2, \qquad \operatorname{Var}\!\left(\|u - u'\|^2\right) = 8(d - r)\sigma_u^4$$

For isotropic $d$-dimensional Gaussian vectors, the relative spread shrinks as $\mathrm{CV}(\|X - X'\|^2) = \sqrt{2/d} \to 0$, so distances concentrate in a thin shell around $\sqrt{2d}$ and the distance contrast
$$C = \frac{D_{\text{far}} - D_{\text{near}}}{D_{\text{near}}}$$
collapses toward $0$ as $d \to \infty$.

---
### 1.3 Same-Class vs. Opposite-Class Squared Distances

Including $z$ with $\delta = (\mu_0 / \sqrt{r})\sigma_z$:
- **Same class ($y = y'$):** $\mathbb{E}\![\|x - x'\|^2 \mid y = y'] = 2r\sigma_z^2 + 2(d - r)\sigma_u^2$
- **Opposite class ($y \neq y'$):** $\mathbb{E}\![\|x - x'\|^2 \mid y \neq y'] = \underbrace{4\mu_0^2 \sigma_z^2}_{\text{between-class gap}} + 2r\sigma_z^2 + 2(d - r)\sigma_u^2$

The resulting **distance signal-to-noise ratio** is:
$$\text{SNR}_{\text{dist}} = \frac{\mathbb{E}[\|x - x'\|^2 \mid y \neq y'] - \mathbb{E}[\|x - x'\|^2 \mid y = y']}{\sqrt{\operatorname{Var}(\|u - u'\|^2)}} = \frac{\sqrt{2}\,\mu_0^2}{\sqrt{d - r}} \left(\frac{\sigma_z}{\sigma_u}\right)^2$$

Whenever $\text{SNR}_{\text{dist}} \ll 1$, random fluctuations from the $d - r$ noise coordinates overwhelm the fixed between-class gap $4\mu_0^2 \sigma_z^2$, pushing KNN and Gaussian KDE toward random guessing ($50\%$).  

---
### 1.4 Regimes of Standardization

Across both classes, the marginal variances are $\operatorname{Var}(u_j) = \sigma_u^2$ and $\operatorname{Var}(z_j) = \sigma_z^2 + \delta^2 = \sigma_z^2(1 + \mu_0^2/r)$. After standardization, the between-class squared-distance gap becomes $\Delta_{\text{std}} = \frac{4\mu_0^2}{1 + \mu_0^2/r}$ while the noise std is $\sqrt{8(d-r)}$, independent of $(\sigma_z, \sigma_u)$:
1. **Noise-Dominated ($\sigma_u \gg \sigma_z$):** Standardization helps dramatically by removing the $100\times$ variance imbalance, but cannot eliminate the $d - r$ unit-variance noise dimensions.
2. **Equal-Scale ($\sigma_u = \sigma_z = 1$):** Standardization cannot help — dividing $z_j$ by $\sqrt{\sigma_z^2 + \delta^2} > \sigma_z$ slightly shrinks signal relative to noise ($\Delta_{\text{std}} \approx 3.35 < 5.76$).
3. **Signal-Dominated ($\sigma_z \gg \sigma_u$):** Standardization actively hurts because it amplifies small noise coordinates up to unit variance.

Only **supervised feature weighting** or **training-only feature selection** removes the $(d - r)$ noise floor and recovers Oracle performance.

---
### 1.5 Why the Four Classifiers Behave Differently
- **KNN & Gaussian KDE (local distance-based):** Rely directly on Euclidean distances in $\mathbb{R}^d$; degrade rapidly with $d$ and large $\sigma_u$, and recover very slowly with $n$.
- **Regularized Gaussian (QDA with isotropic shrinkage):** Shrinks sample covariance $\hat{\Sigma}_c^{\text{reg}} = (1-\alpha)S_c + \alpha \frac{\operatorname{Tr}(S_c)}{d} I_d$, stabilizing estimation in high $d$ when scales are comparable.
- **L2 Logistic Regression:** Learns linear weights $w = (w_z, w_u)$; L2 regularization shrinks irrelevant weights $w_u \to 0$ while preserving $w_z$, making it robust to both high $d$ and unequal scales.

## 3. Synthetic Data Generation & Sanity Check

We generate independent `train`, `val`, and `test` splits using separate NumPy `default_rng` streams (`seed`, `seed + 10_000`, `seed + 20_000`) so there is zero data leakage across splits and exact reproducibility across runs.

```python
def generate_data(
    n_samples,
    d,
    r,
    signal_norm=1.2,
    informative_scale=1.0,
    noise_scale=1.0,
    rng=None,
):
    if rng is None:
        rng = np.random.default_rng()

    # Balanced binary labels: P(y=0) = P(y=1) = 0.5
    n0 = n_samples // 2
    y = np.array([0] * n0 + [1] * (n_samples - n0), dtype=int)
    rng.shuffle(y)

    # Scale per-coordinate shift by 1/sqrt(r) so total SNR (and Bayes error) stays
    # constant across informative dimension r and across rescaling of z
    delta = (signal_norm / np.sqrt(r)) * informative_scale
    sign = (2 * y - 1)[:, None]
    z = sign * delta + rng.normal(0.0, informative_scale, size=(n_samples, r))

    if d > r:
        u = rng.normal(0.0, noise_scale, size=(n_samples, d - r))
        X = np.hstack([z, u])
    else:
        X = z

    return X, y


def generate_splits(
    n_train,
    n_val,
    n_test,
    d,
    r,
    signal_norm=1.2,
    informative_scale=1.0,
    noise_scale=1.0,
    seed=42,
):
    # Separate RNGs so train, val, and test are independent and reproducible
    rng_train = np.random.default_rng(seed)
    rng_val = np.random.default_rng(seed + 10_000)
    rng_test = np.random.default_rng(seed + 20_000)

    X_train, y_train = generate_data(
        n_train, d, r, signal_norm, informative_scale, noise_scale, rng_train
    )
    X_val, y_val = generate_data(
        n_val, d, r, signal_norm, informative_scale, noise_scale, rng_val
    )
    X_test, y_test = generate_data(
        n_test, d, r, signal_norm, informative_scale, noise_scale, rng_test
    )

    return X_train, y_train, X_val, y_val, X_test, y_test


# Sanity check on a sample split (d=50, r=2)
X_tr_demo, y_tr_demo, X_va_demo, y_va_demo, X_te_demo, y_te_demo = generate_splits(
    500, 500, 1000, d=50, r=2, signal_norm=SIGNAL_NORM, seed=42
)
print(f"Shapes        -> train: {X_tr_demo.shape}, val: {X_va_demo.shape}, test: {X_te_demo.shape}")
print(f"Class balance -> train P(y=1)={y_tr_demo.mean():.3f}, val={y_va_demo.mean():.3f}, test={y_te_demo.mean():.3f}")
print(f"Missing/NaN   -> train={np.isnan(X_tr_demo).sum()}, val={np.isnan(X_va_demo).sum()}, test={np.isnan(X_te_demo).sum()}")

# Visualize informative (z1, z2) vs. irrelevant (u1, u2) coordinates
fig, axes = plt.subplots(1, 2, figsize=(11.5, 4.4), dpi=140)
for cls, color, label in [(0, "#1f77b4", "Class 0"), (1, "#d62728", "Class 1")]:
    mask = y_tr_demo == cls
    axes[0].scatter(X_tr_demo[mask, 0], X_tr_demo[mask, 1], alpha=0.55, s=22, color=color, label=label)
    axes[1].scatter(X_tr_demo[mask, 2], X_tr_demo[mask, 3], alpha=0.55, s=22, color=color, label=label)

axes[0].set_title("Informative Coordinates ($z_1, z_2$)")
axes[0].set_xlabel("$z_1$")
axes[0].set_ylabel("$z_2$")
axes[0].legend(frameon=True)

axes[1].set_title("Irrelevant Noise Coordinates ($u_1, u_2$)")
axes[1].set_xlabel("$u_1$")
axes[1].set_ylabel("$u_2$")
axes[1].legend(frameon=True)

fig.suptitle("Synthetic Dataset Preview ($d=50, r=2, \\mu_0=1.2$)", fontweight="bold")
plt.tight_layout()
plt.show()
```

**Data Integrity & Geometry Check:** The generated splits are exactly balanced ($P(y=1) = 0.500$) with zero missing or NaN entries. In the left panel, the two classes separate cleanly along the diagonal $z_1 + z_2 = 0$ with Bayes optimal error $\Phi(-1.2) \approx 11.51\%$; in the right panel, the irrelevant coordinates $(u_1, u_2)$ overlap completely and carry zero class information.

## 4. Core Implementations

### 4.1 Classifiers
1. **`KNNClassifier`**: Euclidean $k$-nearest neighbors with posterior estimated as the fraction of class-1 neighbors.
2. **`GaussianKDEClassifier`**: Class-conditional isotropic Gaussian kernel density estimator evaluated in log-space using `logsumexp` to prevent floating-point underflow in high dimensions.
3. **`GaussianClassifier`**: Regularized class-conditional Gaussian (QDA) with covariance shrinkage $\Sigma_c^{\text{reg}} = (1-\alpha)S_c + \alpha \frac{\operatorname{Tr}(S_c)}{d} I_d + 10^{-6} I_d$.
4. **`LogisticRegressionClassifier`**: L2-regularized logistic regression via `lbfgs`.

```python
class KNNClassifier:
    def __init__(self, n_neighbors=5):
        self.n_neighbors = n_neighbors

    def fit(self, X, y):
        self.X_train = np.asarray(X, dtype=float)
        self.y_train = np.asarray(y, dtype=int)
        return self

    def predict_proba(self, X):
        X = np.asarray(X, dtype=float)
        k = min(self.n_neighbors, len(self.y_train))
        dists = cdist(X, self.X_train, metric="euclidean")
        nn_idx = np.argpartition(dists, kth=k - 1, axis=1)[:, :k]
        p1 = np.mean(self.y_train[nn_idx] == 1, axis=1)
        return np.column_stack([1.0 - p1, p1])

    def predict(self, X):
        return (self.predict_proba(X)[:, 1] >= 0.5).astype(int)


class GaussianKDEClassifier:
    def __init__(self, bandwidth=1.0):
        self.bandwidth = bandwidth

    def fit(self, X, y):
        X = np.asarray(X, dtype=float)
        y = np.asarray(y, dtype=int)
        self.d = X.shape[1]
        self.X_class = [X[y == 0], X[y == 1]]
        self.priors = [len(self.X_class[0]) / len(y), len(self.X_class[1]) / len(y)]
        return self

    def _class_log_density(self, X, c):
        Xc = self.X_class[c]
        h = self.bandwidth
        dist_sq = cdist(X, Xc, metric="sqeuclidean")
        # Compute in log-space to avoid floating-point underflow in high d
        log_kernel_sum = logsumexp(-dist_sq / (2.0 * h * h), axis=1)
        log_norm = -np.log(len(Xc)) - 0.5 * self.d * np.log(2.0 * np.pi) - self.d * np.log(h)
        return log_norm + log_kernel_sum

    def predict_log_proba(self, X):
        X = np.asarray(X, dtype=float)
        log_p0 = self._class_log_density(X, 0) + np.log(self.priors[0])
        log_p1 = self._class_log_density(X, 1) + np.log(self.priors[1])
        joint = np.column_stack([log_p0, log_p1])
        return joint - logsumexp(joint, axis=1, keepdims=True)

    def predict_proba(self, X):
        return np.exp(self.predict_log_proba(X))

    def predict(self, X):
        lp = self.predict_log_proba(X)
        return (lp[:, 1] >= lp[:, 0]).astype(int)


class GaussianClassifier:
    def __init__(self, alpha=0.1):
        self.alpha = alpha

    def fit(self, X, y):
        X = np.asarray(X, dtype=float)
        y = np.asarray(y, dtype=int)
        d = X.shape[1]
        self.priors = [np.mean(y == 0), np.mean(y == 1)]
        self.means = []
        self.covs = []

        for c in (0, 1):
            Xc = X[y == c]
            mu = Xc.mean(axis=0)
            S = np.cov(Xc, rowvar=False) if d > 1 else np.array([[np.var(Xc, ddof=1)]])
            # Shrink sample covariance toward an isotropic diagonal matrix
            avg_var = np.trace(S) / d
            cov = (1.0 - self.alpha) * S + self.alpha * avg_var * np.eye(d) + 1e-6 * np.eye(d)
            self.means.append(mu)
            self.covs.append(cov)
        return self

    def predict_log_proba(self, X):
        X = np.asarray(X, dtype=float)
        log_p0 = multivariate_normal.logpdf(X, mean=self.means[0], cov=self.covs[0]) + np.log(self.priors[0])
        log_p1 = multivariate_normal.logpdf(X, mean=self.means[1], cov=self.covs[1]) + np.log(self.priors[1])
        joint = np.column_stack([log_p0, log_p1])
        return joint - logsumexp(joint, axis=1, keepdims=True)

    def predict_proba(self, X):
        return np.exp(self.predict_log_proba(X))

    def predict(self, X):
        lp = self.predict_log_proba(X)
        return (lp[:, 1] >= lp[:, 0]).astype(int)


class LogisticRegressionClassifier:
    def __init__(self, C=1.0, max_iter=2000):
        self.C = C
        self.model = LogisticRegression(C=C, solver="lbfgs", max_iter=max_iter)

    def fit(self, X, y):
        self.model.fit(X, y)
        return self

    def predict_proba(self, X):
        return self.model.predict_proba(X)

    def predict(self, X):
        return self.model.predict(X).astype(int)
```

### 4.2 Preprocessing Pipelines
All preprocessing transformers (`Standardizer`, `FeatureWeighter`, `FeatureSelector`) are **fit strictly on the training split** before transforming validation and test data.

```python
class RawTransformer:
    def fit(self, X, y=None):
        return self

    def transform(self, X):
        return np.asarray(X, dtype=float)


class Standardizer:
    def __init__(self):
        self.scaler = StandardScaler()

    def fit(self, X, y=None):
        self.scaler.fit(X)
        return self

    def transform(self, X):
        return self.scaler.transform(X)


class FeatureWeighter:
    """Standardize features, then weight each coordinate by its training ANOVA F-score."""

    def __init__(self):
        self.scaler = StandardScaler()
        self.weights = None

    def fit(self, X, y):
        X_std = self.scaler.fit_transform(X)
        f_vals, _ = f_classif(X_std, y)
        f_vals = np.nan_to_num(f_vals, nan=1e-4, posinf=1e4, neginf=1e-4)
        f_vals = np.maximum(f_vals, 1e-4)
        # Normalize weights so average weight is 1
        self.weights = (f_vals / np.sum(f_vals)) * X.shape[1]
        return self

    def transform(self, X):
        return self.scaler.transform(X) * np.sqrt(self.weights)


class FeatureSelector:
    """Training-only top-k feature selection via ANOVA F-score (after standardization)."""

    def __init__(self, k_features=2):
        self.k_features = k_features
        self.scaler = StandardScaler()
        self.selector = SelectKBest(score_func=f_classif, k=k_features)

    def fit(self, X, y):
        X_std = self.scaler.fit_transform(X)
        self.selector.set_params(k=min(self.k_features, X.shape[1]))
        self.selector.fit(X_std, y)
        return self

    def transform(self, X):
        return self.selector.transform(self.scaler.transform(X))


class OracleSelector:
    """Diagnostic baseline using only the true informative coordinates z."""

    def __init__(self, r=2):
        self.r = r

    def fit(self, X, y=None):
        return self

    def transform(self, X):
        return np.asarray(X[:, : self.r], dtype=float)
```

### 4.3 Distance Concentration Metrics, Evaluation, Validation Tuning & Plotting Utilities
- **Metrics**: Classification error rate, predictive log-loss (clipped to $[10^{-15}, 1 - 10^{-15}]$), mean nearest-neighbor distance $D_{\text{near}}$, mean farthest-neighbor distance $D_{\text{far}}$, contrast $C = (D_{\text{far}} - D_{\text{near}}) / D_{\text{near}}$, and ratio $R = D_{\text{near}} / D_{\text{far}}$.
- **Validation tuning**: Probabilistic classifiers (`GaussianKDE`, `Gaussian`, `LogisticRegression`) are tuned by validation log-loss; `KNN` is tuned by validation error rate with log-loss tie-breaking.

```python
CLASSIFIERS = ["KNN", "GaussianKDE", "Gaussian", "LogisticRegression"]

PARAM_GRIDS = {
    "KNN": [1, 3, 5, 7, 11, 15, 21, 31, 51],
    "GaussianKDE": [0.2, 0.5, 1.0, 2.0, 3.0, 5.0, 10.0, 20.0, 50.0, 100.0],
    "Gaussian": [0.0, 0.01, 0.05, 0.1, 0.2, 0.5, 0.8, 1.0],
    "LogisticRegression": [1e-4, 1e-3, 1e-2, 1e-1, 1.0, 10.0, 100.0],
}


def compute_distance_metrics(X_test, X_train, eps=1e-12):
    dists = cdist(X_test, X_train, metric="euclidean")
    d_near = np.min(dists, axis=1)
    d_far = np.max(dists, axis=1)

    contrast = (d_far - d_near) / np.maximum(d_near, eps)
    ratio = d_near / np.maximum(d_far, eps)

    return {
        "mean_d_near": float(np.mean(d_near)),
        "mean_d_far": float(np.mean(d_far)),
        "mean_contrast": float(np.mean(contrast)),
        "mean_ratio": float(np.mean(ratio)),
    }


def error_rate(y_true, y_pred):
    return float(np.mean(np.asarray(y_true) != np.asarray(y_pred)))


def log_loss(y_true, y_prob, eps=1e-15):
    y_true = np.asarray(y_true, dtype=float)
    y_prob = np.asarray(y_prob, dtype=float)
    p1 = y_prob[:, 1] if y_prob.ndim == 2 else y_prob
    p1 = np.clip(p1, eps, 1.0 - eps)
    return float(-np.mean(y_true * np.log(p1) + (1.0 - y_true) * np.log(1.0 - p1)))


def make_classifier(name, param):
    if name == "KNN":
        return KNNClassifier(n_neighbors=int(param))
    if name == "GaussianKDE":
        return GaussianKDEClassifier(bandwidth=float(param))
    if name == "Gaussian":
        return GaussianClassifier(alpha=float(param))
    if name == "LogisticRegression":
        return LogisticRegressionClassifier(C=float(param))
    raise ValueError(f"Unknown classifier: {name}")


def tune_classifier(name, X_train, y_train, X_val, y_val):
    best_model = None
    best_param = None
    best_val_err = float("inf")
    best_val_loss = float("inf")

    for val in PARAM_GRIDS[name]:
        param = min(int(val), len(y_train)) if name == "KNN" else val
        model = make_classifier(name, param).fit(X_train, y_train)

        val_err = error_rate(y_val, model.predict(X_val))
        val_loss = log_loss(y_val, model.predict_proba(X_val))

        if name == "KNN":
            better = (val_err < best_val_err - 1e-9) or (
                abs(val_err - best_val_err) <= 1e-9 and val_loss < best_val_loss
            )
        else:
            better = val_loss < best_val_loss

        if better:
            best_model = model
            best_param = param
            best_val_err = val_err
            best_val_loss = val_loss

    return best_model, best_param, best_val_err, best_val_loss


def evaluate_classifier(name, X_train, y_train, X_val, y_val, X_test, y_test, dist_metrics=None):
    model, best_param, val_err, val_loss = tune_classifier(name, X_train, y_train, X_val, y_val)

    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)

    if dist_metrics is None:
        dist_metrics = compute_distance_metrics(X_test, X_train)

    return {
        "classifier": name,
        "error_rate": error_rate(y_test, y_pred),
        "log_loss": log_loss(y_test, y_prob),
        "best_param": best_param,
        "val_error": val_err,
        "val_loss": val_loss,
        **dist_metrics,
    }


def summarize_results(raw_df, group_cols):
    metrics = ["error_rate", "log_loss", "mean_d_near", "mean_d_far", "mean_contrast", "mean_ratio"]
    agg = {}
    for col in metrics:
        agg[f"{col}_mean"] = (col, "mean")
        agg[f"{col}_std"] = (col, "std")
    return raw_df.groupby(group_cols).agg(**agg).reset_index()


def save_results(raw_df, summary_df, exp_name, results_dir=RESULTS_DIR):
    if not SAVE_ARTIFACTS:
        return
    csv_dir = os.path.join(results_dir, "csv")
    tables_dir = os.path.join(results_dir, "tables")
    os.makedirs(csv_dir, exist_ok=True)
    os.makedirs(tables_dir, exist_ok=True)

    raw_df.to_csv(os.path.join(csv_dir, f"{exp_name}_raw.csv"), index=False)
    summary_df.to_csv(os.path.join(csv_dir, f"{exp_name}_summary.csv"), index=False)

    md = tabulate(summary_df.round(4), headers="keys", tablefmt="pipe", showindex=False)
    with open(os.path.join(tables_dir, f"{exp_name}_table.md"), "w", encoding="utf-8") as f:
        f.write(f"# {exp_name}\n\n{md}\n")


def format_compact_table(summary_df, index_cols):
    out = summary_df[index_cols].copy()
    out["Error (Mean ± Std)"] = summary_df.apply(
        lambda row: f"{row['error_rate_mean']:.4f} ± {row['error_rate_std']:.4f}", axis=1
    )
    out["Log-Loss (Mean ± Std)"] = summary_df.apply(
        lambda row: f"{row['log_loss_mean']:.4f} ± {row['log_loss_std']:.4f}", axis=1
    )
    out["D_near"] = summary_df["mean_d_near_mean"].round(3)
    out["D_far"] = summary_df["mean_d_far_mean"].round(3)
    out["Contrast C"] = summary_df["mean_contrast_mean"].round(3)
    return out
```

```python
def plot_curves(
    summary_df,
    x_col,
    y_mean,
    y_std,
    x_label,
    y_label,
    title,
    save_path=None,
    log_x=False,
    ylim=None,
    hline=None,
    hline_label=None,
):
    fig, ax = plt.subplots(figsize=(7.8, 4.8), dpi=150)

    for clf, group in summary_df.groupby("classifier"):
        g = group.sort_values(x_col)
        x = g[x_col].values
        mean = g[y_mean].values
        std = g[y_std].values
        c = COLORS.get(clf)
        ax.plot(x, mean, marker=MARKERS.get(clf, "o"), label=clf, color=c, linewidth=2)
        ax.fill_between(x, mean - std, mean + std, color=c, alpha=0.16)

    if hline is not None:
        ax.axhline(hline, color="black", linestyle="--", linewidth=1.2, label=hline_label)

    if log_x:
        ax.set_xscale("log")
    if ylim is not None:
        ax.set_ylim(ylim)

    ax.set_xlabel(x_label)
    ax.set_ylabel(y_label)
    ax.set_title(title)
    ax.legend(frameon=True)
    plt.tight_layout()
    if SAVE_ARTIFACTS and save_path:
        os.makedirs(os.path.dirname(os.path.abspath(save_path)), exist_ok=True)
        fig.savefig(save_path, dpi=200)
    plt.show()


def plot_distance_summary(summary_df, x_col, x_label, title, save_path=None, log_x=False):
    fig, axes = plt.subplots(1, 3, figsize=(14.5, 4.3), dpi=150)

    sub = summary_df[summary_df["classifier"] == summary_df["classifier"].iloc[0]].sort_values(x_col)
    x = sub[x_col].values

    panels = [
        ("mean_d_near", "Mean Nearest Distance ($D_{near}$)", "Nearest Distance", "#1f77b4", "o"),
        ("mean_d_far", "Mean Farthest Distance ($D_{far}$)", "Farthest Distance", "#ff7f0e", "s"),
        ("mean_contrast", "Contrast $(D_{far} - D_{near}) / D_{near}$", "Distance Contrast", "#2ca02c", "^"),
    ]
    for ax, (col, ylabel, ptitle, color, marker) in zip(axes, panels):
        m = sub[f"{col}_mean"].values
        s = sub[f"{col}_std"].values
        ax.plot(x, m, marker=marker, color=color, linewidth=2)
        ax.fill_between(x, m - s, m + s, color=color, alpha=0.18)
        if log_x:
            ax.set_xscale("log")
        ax.set_xlabel(x_label)
        ax.set_ylabel(ylabel)
        ax.set_title(ptitle)

    fig.suptitle(title, fontweight="bold")
    plt.tight_layout()
    if SAVE_ARTIFACTS and save_path:
        os.makedirs(os.path.dirname(os.path.abspath(save_path)), exist_ok=True)
        fig.savefig(save_path, dpi=200)
    plt.show()


def plot_scale_regimes(summary_df, metric_mean, metric_std, y_label, title, save_path=None):
    regimes = ["Noise-Dominated", "Equal-Scale", "Signal-Dominated"]
    settings = ["Raw", "Standardized", "Weighted", "Selected", "Oracle"]
    classifiers = ["KNN", "GaussianKDE", "Gaussian", "LogisticRegression"]
    palette = sns.color_palette("muted", len(settings))

    fig, axes = plt.subplots(1, 3, figsize=(16, 4.8), dpi=150, sharey=True)
    x = np.arange(len(classifiers))
    width = 0.8 / len(settings)

    for ax, regime in zip(axes, regimes):
        rdf = summary_df[summary_df["regime"] == regime]
        for i, setting in enumerate(settings):
            sdf = rdf[rdf["setting"] == setting].set_index("classifier")
            means = [sdf.loc[c, metric_mean] for c in classifiers]
            stds = [sdf.loc[c, metric_std] for c in classifiers]
            offset = (i - len(settings) / 2 + 0.5) * width
            ax.bar(
                x + offset,
                means,
                width=width,
                yerr=stds,
                label=setting,
                color=palette[i],
                capsize=2.5,
                edgecolor="black",
                linewidth=0.6,
            )
        ax.set_xticks(x)
        ax.set_xticklabels(["KNN", "KDE", "Gaussian", "LogReg"])
        ax.set_title(regime)

    axes[0].set_ylabel(y_label)
    axes[-1].legend(title="Pipeline", loc="upper right", fontsize=8.5)
    fig.suptitle(title, fontweight="bold")
    plt.tight_layout()
    if SAVE_ARTIFACTS and save_path:
        os.makedirs(os.path.dirname(os.path.abspath(save_path)), exist_ok=True)
        fig.savefig(save_path, dpi=200)
    plt.show()


def plot_paired_informative_dim(summary_df, y_mean, y_std, y_label, title, save_path=None, hline=None):
    modes = ["Fixed Ambient (d=50)", "No Noise (d=r)"]
    fig, axes = plt.subplots(1, 2, figsize=(13.5, 4.7), dpi=150, sharey=True)

    for ax, mode in zip(axes, modes):
        mdf = summary_df[summary_df["mode"] == mode]
        for clf, group in mdf.groupby("classifier"):
            g = group.sort_values("r")
            x = g["r"].values
            m = g[y_mean].values
            s = g[y_std].values
            c = COLORS.get(clf)
            ax.plot(x, m, marker=MARKERS.get(clf, "o"), label=clf, color=c, linewidth=2)
            ax.fill_between(x, m - s, m + s, color=c, alpha=0.16)
        if hline is not None:
            ax.axhline(hline, color="black", linestyle="--", linewidth=1.2, label="Bayes Optimal")
        ax.set_xlabel("Informative Dimension ($r$)")
        ax.set_title(mode)
        ax.legend(frameon=True, fontsize=8.5)

    axes[0].set_ylabel(y_label)
    fig.suptitle(title, fontweight="bold")
    plt.tight_layout()
    if SAVE_ARTIFACTS and save_path:
        os.makedirs(os.path.dirname(os.path.abspath(save_path)), exist_ok=True)
        fig.savefig(save_path, dpi=200)
    plt.show()
```

## 5. Experiment 1: Increasing Ambient Dimension ($d$)

- **Fixed:** $r = 2$, $n_{\text{train}} = 500$, $n_{\text{val}} = 500$, $n_{\text{test}} = 1000$, $\sigma_z = \sigma_u = 1.0$, $\mu_0 = 1.2$.
- **Varied:** $d \in \{2, 5, 10, 20, 50, 100, 200\}$ across 5 random seeds (42–46).

This experiment directly measures how adding independent Gaussian noise coordinates collapses distance contrast $C$ and degrades local vs. parametric classifiers.

```python
t0 = time.time()
d_list = [2, 5, 10, 20, 50, 100, 200]
r_exp1 = 2
n_train_1, n_val_1, n_test_1 = 500, 500, 1000

rows_exp1 = []
for d in d_list:
    for s in range(N_SEEDS):
        seed = 42 + s
        X_tr, y_tr, X_va, y_va, X_te, y_te = generate_splits(
            n_train_1, n_val_1, n_test_1, d=d, r=r_exp1, signal_norm=SIGNAL_NORM, seed=seed
        )
        dist_metrics = compute_distance_metrics(X_te, X_tr)

        for clf in CLASSIFIERS:
            res = evaluate_classifier(clf, X_tr, y_tr, X_va, y_va, X_te, y_te, dist_metrics)
            res.update({"d": d, "r": r_exp1, "n_train": n_train_1, "seed": seed})
            rows_exp1.append(res)

exp1_raw = pd.DataFrame(rows_exp1)
exp1_summary = summarize_results(exp1_raw, ["d", "classifier"])
save_results(exp1_raw, exp1_summary, "experiment_1_ambient_dimension", RESULTS_DIR)

print(f"Experiment 1 finished in {time.time() - t0:.1f}s")
display(format_compact_table(exp1_summary, ["d", "classifier"]))
```

```python
plots_dir = os.path.join(RESULTS_DIR, "plots")

plot_curves(
    exp1_summary, "d", "error_rate_mean", "error_rate_std",
    x_label="Ambient Dimension ($d$)",
    y_label="Classification Error Rate",
    title="Exp 1: Error Rate vs Ambient Dimension ($r=2, n=500$)",
    save_path=os.path.join(plots_dir, "exp1_error_vs_ambient_dim.png"),
    hline=BAYES_ERROR, hline_label=f"Bayes Optimal ({BAYES_ERROR:.3f})",
)

plot_curves(
    exp1_summary, "d", "log_loss_mean", "log_loss_std",
    x_label="Ambient Dimension ($d$)",
    y_label="Predictive Log-Loss",
    title="Exp 1: Predictive Log-Loss vs Ambient Dimension ($r=2, n=500$)",
    save_path=os.path.join(plots_dir, "exp1_log_loss_vs_ambient_dim.png"),
)

plot_distance_summary(
    exp1_summary, "d",
    x_label="Ambient Dimension ($d$)",
    title="Exp 1: Distance Concentration vs Ambient Dimension",
    save_path=os.path.join(plots_dir, "exp1_distance_concentration.png"),
)
```

### Experiment 1 Analysis
- **Distance concentration matches theory:** As $d$ increases from $2$ to $200$, $D_{\text{near}}$ climbs from $0.130$ to $17.430$ (tracking $\sqrt{2d}$) while distance contrast $C$ collapses from **$84.15$ down to $0.305$** — the farthest training point is only $30\%$ farther than the nearest neighbor.
- **Local distance-based classifiers degrade fastest:** At $d=2$, all four models match the Bayes optimal error ($11.5\%$). By $d=200$, KNN rises to **$21.68\%$** error (log-loss $0.569$) and Gaussian KDE rises to **$22.98\%$** error (log-loss $0.574$).
- **Parametric models stay markedly more resilient:** Regularized Gaussian ($14.00\%$ error, log-loss $0.336$) and L2 Logistic Regression ($15.34\%$ error, log-loss $0.396$) suppress irrelevant coordinates through covariance shrinkage and L2 weight decay.
- *Note on KNN log-loss at $d=2$:* Because raw KNN probabilities are neighbor proportions $m/k$, small $k$ occasionally predicts $0$ or $1$, causing elevated clipped log-loss variance despite low classification error.

## 6. Experiment 2: Increasing Sample Size ($n$)

- **Fixed:** $r = 2$, $d = 50$, $n_{\text{val}} = 500$, $n_{\text{test}} = 1000$, $\sigma_z = \sigma_u = 1.0$, $\mu_0 = 1.2$.
- **Varied:** $n_{\text{train}} \in \{100, 250, 500, 1000, 2500, 5000\}$ across 5 random seeds (42–46).

This experiment tests whether scaling up the training set by $50\times$ can overcome 48 irrelevant dimensions for local vs. parametric classifiers.

```python
t0 = time.time()
n_list = [100, 250, 500, 1000, 2500, 5000]
d_exp2, r_exp2 = 50, 2
n_val_2, n_test_2 = 500, 1000

rows_exp2 = []
for n_train in n_list:
    for s in range(N_SEEDS):
        seed = 42 + s
        X_tr, y_tr, X_va, y_va, X_te, y_te = generate_splits(
            n_train, n_val_2, n_test_2, d=d_exp2, r=r_exp2, signal_norm=SIGNAL_NORM, seed=seed
        )
        dist_metrics = compute_distance_metrics(X_te, X_tr)

        for clf in CLASSIFIERS:
            res = evaluate_classifier(clf, X_tr, y_tr, X_va, y_va, X_te, y_te, dist_metrics)
            res.update({"n_train": n_train, "d": d_exp2, "r": r_exp2, "seed": seed})
            rows_exp2.append(res)

exp2_raw = pd.DataFrame(rows_exp2)
exp2_summary = summarize_results(exp2_raw, ["n_train", "classifier"])
save_results(exp2_raw, exp2_summary, "experiment_2_sample_size", RESULTS_DIR)

print(f"Experiment 2 finished in {time.time() - t0:.1f}s")
display(format_compact_table(exp2_summary, ["n_train", "classifier"]))
```

```python
plot_curves(
    exp2_summary, "n_train", "error_rate_mean", "error_rate_std",
    x_label="Training Sample Size ($n$)",
    y_label="Classification Error Rate",
    title="Exp 2: Error Rate vs Sample Size ($d=50, r=2$)",
    save_path=os.path.join(plots_dir, "exp2_error_vs_sample_size.png"),
    log_x=True,
    hline=BAYES_ERROR, hline_label=f"Bayes Optimal ({BAYES_ERROR:.3f})",
)

plot_curves(
    exp2_summary, "n_train", "log_loss_mean", "log_loss_std",
    x_label="Training Sample Size ($n$)",
    y_label="Predictive Log-Loss",
    title="Exp 2: Predictive Log-Loss vs Sample Size ($d=50, r=2$)",
    save_path=os.path.join(plots_dir, "exp2_log_loss_vs_sample_size.png"),
    log_x=True,
)

plot_distance_summary(
    exp2_summary, "n_train",
    x_label="Training Sample Size ($n$)",
    title="Exp 2: Distance Concentration vs Sample Size",
    save_path=os.path.join(plots_dir, "exp2_distance_vs_sample_size.png"),
    log_x=True,
)
```

### Experiment 2 Analysis
- **Parametric models converge to the Bayes floor:** Increasing $n_{\text{train}}$ by **$50\times$** (from $100$ to $5000$) drives Regularized Gaussian from $15.04\%$ down to **$11.46\%$** (log-loss $0.272$) and L2 Logistic Regression from $16.26\%$ down to **$11.74\%$** (log-loss $0.275$).
- **Local geometry barely improves with $50\times$ more data:** Because filling a $50$-dimensional ball requires $n$ exponential in $d$, $D_{\text{near}}$ drops only slightly from $7.895$ to $6.946$ and contrast $C$ remains below $1.0$ ($0.568 \to 0.938$).
- **Persistent gap for KNN and KDE:** Even at $n_{\text{train}} = 5000$, KNN stays stuck at **$14.02\%$** error (log-loss $0.405$) and Gaussian KDE log-loss remains flat at **$0.472$**.

## 7. Experiment 3: Unequal Feature Scales & Standardization Regimes

- **Fixed:** $r = 2$, $d = 50$, $n_{\text{train}} = 500$, $n_{\text{val}} = 500$, $n_{\text{test}} = 1000$, constant Bayes error $11.51\%$.
- **Three feature-scale regimes:**
  1. **Noise-Dominated** ($\sigma_z = 1.0, \sigma_u = 10.0$)
  2. **Equal-Scale** ($\sigma_z = 1.0, \sigma_u = 1.0$)
  3. **Signal-Dominated** ($\sigma_z = 10.0, \sigma_u = 1.0$)
- **Five preprocessing pipelines** (all fit on training data only): **Raw**, **Standardized**, **Weighted** (standardize + training ANOVA $F$-score weights), **Selected** (training-only ANOVA `SelectKBest` with $k \in \{1, 2, 3, 5, 10, 20, 50\}$ tuned on validation), and **Oracle** (true informative $z$ only).

```python
def run_feature_selection_pipeline(clf, X_tr, y_tr, X_va, y_va, X_te, y_te, k_grid):
    best_model, best_sel, best_k, best_param = None, None, None, None
    best_val_err, best_val_loss = float("inf"), float("inf")

    for k in k_grid:
        sel = FeatureSelector(k_features=k).fit(X_tr, y_tr)
        model, param, val_err, val_loss = tune_classifier(
            clf, sel.transform(X_tr), y_tr, sel.transform(X_va), y_va
        )
        if clf == "KNN":
            better = (val_err < best_val_err - 1e-9) or (
                abs(val_err - best_val_err) <= 1e-9 and val_loss < best_val_loss
            )
        else:
            better = val_loss < best_val_loss

        if better:
            best_model, best_sel, best_k, best_param = model, sel, k, param
            best_val_err, best_val_loss = val_err, val_loss

    X_tr_sel = best_sel.transform(X_tr)
    X_te_sel = best_sel.transform(X_te)
    dist_metrics = compute_distance_metrics(X_te_sel, X_tr_sel)

    return {
        "classifier": clf,
        "error_rate": error_rate(y_te, best_model.predict(X_te_sel)),
        "log_loss": log_loss(y_te, best_model.predict_proba(X_te_sel)),
        "best_param": f"{best_param}_k{best_k}",
        "val_error": best_val_err,
        "val_loss": best_val_loss,
        **dist_metrics,
    }


t0 = time.time()
d_exp3, r_exp3 = 50, 2
n_train_3, n_val_3, n_test_3 = 500, 500, 1000
k_grid_3 = [1, 2, 3, 5, 10, 20, 50]

regimes = [
    ("Noise-Dominated", 1.0, 10.0),
    ("Equal-Scale", 1.0, 1.0),
    ("Signal-Dominated", 10.0, 1.0),
]

rows_exp3 = []
for regime_name, sig_z, sig_u in regimes:
    for s in range(N_SEEDS):
        seed = 42 + s
        X_tr, y_tr, X_va, y_va, X_te, y_te = generate_splits(
            n_train_3, n_val_3, n_test_3, d=d_exp3, r=r_exp3,
            signal_norm=SIGNAL_NORM,
            informative_scale=sig_z,
            noise_scale=sig_u,
            seed=seed,
        )

        pipelines = {
            "Raw": RawTransformer().fit(X_tr, y_tr),
            "Standardized": Standardizer().fit(X_tr, y_tr),
            "Weighted": FeatureWeighter().fit(X_tr, y_tr),
            "Oracle": OracleSelector(r=r_exp3).fit(X_tr, y_tr),
        }

        for setting, trans in pipelines.items():
            Xt_tr = trans.transform(X_tr)
            Xt_va = trans.transform(X_va)
            Xt_te = trans.transform(X_te)
            dist_metrics = compute_distance_metrics(Xt_te, Xt_tr)

            for clf in CLASSIFIERS:
                res = evaluate_classifier(clf, Xt_tr, y_tr, Xt_va, y_va, Xt_te, y_te, dist_metrics)
                res.update({"regime": regime_name, "setting": setting, "seed": seed})
                rows_exp3.append(res)

        for clf in CLASSIFIERS:
            res = run_feature_selection_pipeline(clf, X_tr, y_tr, X_va, y_va, X_te, y_te, k_grid_3)
            res.update({"regime": regime_name, "setting": "Selected", "seed": seed})
            rows_exp3.append(res)

exp3_raw = pd.DataFrame(rows_exp3)
exp3_summary = summarize_results(exp3_raw, ["regime", "setting", "classifier"])
save_results(exp3_raw, exp3_summary, "experiment_3_feature_scales", RESULTS_DIR)

print(f"Experiment 3 finished in {time.time() - t0:.1f}s")

# Compact pivot table of mean classification error rate
regime_order = ["Noise-Dominated", "Equal-Scale", "Signal-Dominated"]
setting_order = ["Raw", "Standardized", "Weighted", "Selected", "Oracle"]
exp3_pivot = exp3_summary.pivot(
    index=["regime", "setting"], columns="classifier", values="error_rate_mean"
).reindex(pd.MultiIndex.from_product([regime_order, setting_order], names=["regime", "setting"]))[CLASSIFIERS]
display(exp3_pivot.round(4))
```

### Experiment 3 Analysis
The empirical results mirror the three standardization regimes derived in Section 1.4:
1. **Noise-Dominated ($\sigma_z = 1, \sigma_u = 10$) — Standardization helps, but cannot remove noise:** Raw KNN ($49.06\%$), KDE ($48.52\%$), and isotropic-shrunk Gaussian ($47.64\%$) collapse to coin-flipping because the 48 noise variables carry $100\times$ larger variance. Standardization cuts KNN/KDE error by more than half ($18.52\%$ / $20.48\%$) and restores Gaussian to $13.90\%$, yet all 48 unit-variance noise features remain.
2. **Equal-Scale ($\sigma_z = 1, \sigma_u = 1$) — Standardization cannot help:** Dividing $z_j$ by its marginal standard deviation $\sqrt{\sigma_z^2 + \delta^2} \approx 1.31$ slightly reduces the between-class distance gap relative to unit-variance noise, nudging KNN error *up* ($14.72\% \to 18.52\%$) and KDE error *up* ($16.02\% \to 20.48\%$).
3. **Signal-Dominated ($\sigma_z = 10, \sigma_u = 1$) — Standardization actively hurts:** Raw KNN ($11.72\%$) and Raw KDE ($11.44\%$) naturally match Oracle because the informative coordinates dominate Euclidean distance. Standardization scales up the 48 tiny noise coordinates to variance 1, degrading KNN to $18.52\%$ and KDE to $20.48\%$.
4. **Supervised weighting and selection recover Oracle in every regime:** Across all three regimes, training-only `Weighted` and `Selected` pipelines achieve **$11.32\%$–$12.10\%$** error (and restore distance contrast from $C \approx 0.71$ up to $8.74$ and $56.3$–$84.4$), matching the Oracle baseline.

## 8. Experiment 4: Paired Isolation of Informative Dimension ($r$) vs. Ambient Dimension ($d$)

- **Fixed:** $n_{\text{train}} = 500$, $n_{\text{val}} = 500$, $n_{\text{test}} = 1000$, constant total signal norm $\mu_0 = 1.2$ ($\delta = 1.2/\sqrt{r}$, constant Bayes optimal error $11.51\%$) across 5 random seeds (42–46).
- **Varied:** $r \in \{1, 2, 4, 8, 16, 32\}$ in two paired settings:
  1. **Fixed Ambient ($d = 50$):** Total dimensionality stays fixed while more coordinates carry signal.
  2. **No Noise ($d = r$):** Zero irrelevant variables, testing whether spreading the same signal across more informative dimensions still triggers the curse of dimensionality.

```python
t0 = time.time()
r_list = [1, 2, 4, 8, 16, 32]
n_train_4, n_val_4, n_test_4 = 500, 500, 1000

modes = [
    ("Fixed Ambient (d=50)", lambda r: 50),
    ("No Noise (d=r)", lambda r: r),
]

rows_exp4 = []
for mode_name, get_d in modes:
    for r in r_list:
        d = get_d(r)
        for s in range(N_SEEDS):
            seed = 42 + s
            X_tr, y_tr, X_va, y_va, X_te, y_te = generate_splits(
                n_train_4, n_val_4, n_test_4, d=d, r=r, signal_norm=SIGNAL_NORM, seed=seed
            )
            dist_metrics = compute_distance_metrics(X_te, X_tr)

            for clf in CLASSIFIERS:
                res = evaluate_classifier(clf, X_tr, y_tr, X_va, y_va, X_te, y_te, dist_metrics)
                res.update({"mode": mode_name, "r": r, "d": d, "seed": seed})
                rows_exp4.append(res)

exp4_raw = pd.DataFrame(rows_exp4)
exp4_summary = summarize_results(exp4_raw, ["mode", "r", "classifier"])
save_results(exp4_raw, exp4_summary, "experiment_4_informative_dimension", RESULTS_DIR)

print(f"Experiment 4 finished in {time.time() - t0:.1f}s")
display(format_compact_table(exp4_summary, ["mode", "r", "classifier"]))
```

```python
plot_paired_informative_dim(
    exp4_summary, "error_rate_mean", "error_rate_std",
    y_label="Classification Error Rate",
    title="Exp 4: Error Rate vs Informative Dimension (Constant Bayes Error)",
    save_path=os.path.join(plots_dir, "exp4_error_vs_informative_dim.png"),
    hline=BAYES_ERROR,
)

plot_paired_informative_dim(
    exp4_summary, "log_loss_mean", "log_loss_std",
    y_label="Predictive Log-Loss",
    title="Exp 4: Predictive Log-Loss vs Informative Dimension (Constant Bayes Error)",
    save_path=os.path.join(plots_dir, "exp4_log_loss_vs_informative_dim.png"),
)
```

### Experiment 4 Analysis
- **Fixed Ambient ($d = 50$):** Because total dimensionality $d = 50$ and total class separation $2\mu_0 = 2.4$ are both constant, the squared-distance distribution barely changes ($C \approx 0.72$ across all $r$) and error rates remain roughly flat from $r=1$ to $r=32$ (KNN $\approx 15\%$, KDE $\approx 16\%$, Gaussian $\approx 12\%$, LogReg $\approx 13\%$).
- **No Noise ($d = r$):** Even with **zero irrelevant variables**, spreading the same fixed signal across $r = 1 \to 32$ informative coordinates causes distance contrast to drop from $10017.0$ to $0.987$, KNN error to climb from $11.94\%$ to $14.22\%$, and KDE log-loss to rise from $0.283$ to $0.472$.
- **Unified takeaway:** For distance-based classifiers, performance degradation is governed by **total dimensionality $d$** relative to sample size $n$ and total separation $\mu_0$ — whether the added dimensions carry pure noise (Exp 1) or dilute fixed signal across many informative coordinates (Exp 4, $d=r$).

## 9. Final Summary

### Q&A
- **How do irrelevant variables affect distance geometry and classification?** Adding independent noise dimensions drives pairwise squared distances into a thin shell around $2(d-r)\sigma_u^2$ with random fluctuation $\sqrt{8(d-r)}\sigma_u^2$. Local distance-based classifiers (KNN, Gaussian KDE) degrade rapidly as distance contrast $C$ collapses, while parametric classifiers (Regularized Gaussian, L2 Logistic Regression) remain far more resilient.
- **Does collecting more training data fix the curse of dimensionality for local methods?** Only very slowly. Scaling $n_{\text{train}}$ by $50\times$ ($100 \to 5000$) at $d=50$ brings Regularized Gaussian and L2 Logistic Regression to the Bayes floor ($11.5\%$), whereas KNN remains at $14.0\%$ error and KDE log-loss remains at $0.472$.
- **When does standardization help, fail, or hurt?** Standardization rescues Noise-Dominated features ($\sigma_u \gg \sigma_z$) from $49\%$ coin-flipping error down to $\approx 19\%$, cannot eliminate the $d-r$ irrelevant dimensions in Equal-Scale data (nudging KNN up from $14.7\%$ to $18.5\%$), and actively hurts Signal-Dominated data ($\sigma_z \gg \sigma_u$, pushing KNN from $11.7\%$ to $18.5\%$). Only supervised feature weighting or feature selection recovers Oracle performance.
- **Is degradation caused solely by irrelevant noise, or by dimensionality itself?** Experiment 4 proves that even with zero irrelevant coordinates ($d = r$) at constant Bayes error ($11.51\%$), increasing dimensionality from $r=1$ to $r=32$ collapses distance contrast ($10017 \to 0.99$) and degrades KNN and KDE.

### Data Analysis Key Findings
- **Ambient dimension (Exp 1, $r=2, n=500$):** Increasing $d$ from $2$ to $200$ increases $D_{\text{near}}$ from $0.130$ to $17.430$ and collapses contrast $C$ from $84.15$ to $0.305$. Error rates at $d=200$ reach $21.68\%$ (KNN) and $22.98\%$ (KDE), compared to $14.00\%$ (Regularized Gaussian) and $15.34\%$ (L2 Logistic Regression) against a $11.51\%$ Bayes floor.
- **Sample size (Exp 2, $d=50, r=2$):** Increasing $n_{\text{train}}$ from $100$ to $5000$ reduces Regularized Gaussian error from $15.04\%$ to $11.46\%$ and Logistic Regression from $16.26\%$ to $11.74\%$, while $D_{\text{near}}$ barely shifts ($7.895 \to 6.946$) and KNN stays at $14.02\%$.
- **Feature scales & preprocessing (Exp 3, $d=50, r=2, n=500$):** Raw KNN error is $49.06\%$ (Noise-Dominated), $14.72\%$ (Equal-Scale), and $11.72\%$ (Signal-Dominated); Standardization maps all three to $18.52\%$. Training-only ANOVA `Weighted` ($11.40\%$–$13.38\%$) and `Selected` ($11.32\%$–$12.08\%$) match the `Oracle` baseline ($11.32\%$–$11.74\%$) across every regime.
- **Paired informative dimension (Exp 4, constant $\mu_0=1.2$):** Under Fixed Ambient ($d=50$), contrast ($C \approx 0.72$) and error remain nearly flat for $r \in [1, 32]$. Under No Noise ($d=r$), contrast drops from $10017.0$ to $0.987$, KNN error rises from $11.94\%$ to $14.22\%$, and KDE log-loss rises from $0.283$ to $0.472$.

### Insights or Next Steps
- Always pair distance-based models (KNN, KDE) with **training-only supervised feature selection or metric learning / feature weighting** rather than blind standardization when irrelevant coordinates may be present.
- As a natural extension, test **correlated noise features**, **L1-regularized logistic regression (sparse feature selection)**, or **Mahalanobis / learned metric KNN** to compare sparse parametric vs. adaptive distance-based recovery in high ambient dimensions.
